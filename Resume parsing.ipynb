```python
import os
from PyPDF2 import PdfReader
```

```python
def pdftotext(pdffile):
    # creating a pdf reader object
    pdfReader = PdfReader(pdffile)
    # printing number of pages in pdf file
    pageContents = []
    number_of_pages = len(pdfReader.pages)
    for i in range(0, number_of_pages):
        page = pdfReader.pages[i]
        pageContents.append(page.extract_text())

    full_text = "\n".join(pageContents)
    return full_text
```

```python
pdffile = "./data/resumes/Profile.pdf"
```

```python
text = pdftotext(pdffile)
print(text)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(text)
```

```python
inputs = text.split("\n")

import re

def text_cleanup(line: str):
    # whitespace cleanup
    return re.sub("\s+", " ", line).strip().lower()
```

```python
cleaned_inputs = filter(lambda x: len(x) != 0, map(text_cleanup, inputs))
docs = list(map(lambda line: nlp(line), cleaned_inputs))
```

```python
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
from spacy.matcher import Matcher

time_matcher = Matcher(nlp.vocab)

oclock = [
    {
        "IS_DIGIT": True
    },
    {
        "TEXT": "o'clock"
    }
]

morning = [
    {
        "LOWER": { "IN" : ["am", "morning"]}
    }
]

time_matcher.add("alert_time", [oclock])
```

```python
def process_input_doc(doc: spacy.tokens.doc.Doc):
    obj = { "time": None, "m": None }

    for ent in doc.ents:
        if ent.label_ == "TIME":
            print(ent)
        elif ent.label == "DATE":
            print(ent.text)
    return obj
```

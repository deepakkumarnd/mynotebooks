```python
from spacy.matcher import Matcher

matcher = Matcher(nlp.vocab)

start_pattern = [{"POS": "VERB", "LOWER": {"NOT_IN": ["to"]}, "IS_SENT_START": True}]
end_pattern = [{"TEXT": "("}, {"TEXT": "pull"}, {"TEXT": "request"}, {"TEXT": "#"}, {"POS": "NUM"}, {"TEXT": ")", "SPACY": False}]
matcher.add("start with verb", [start_pattern, end_pattern])

def isGoodCommitMessage(message: str) -> bool:
    doc = nlp(message)
    matches = matcher(doc)
    
#     for token in doc:
#         print(token.text, token.pos_)
        
    for match_id, start, end in matches:
        print(doc[start:end])
    
    if len(matches) == 2:
        return True
    else: 
        return False
    
```

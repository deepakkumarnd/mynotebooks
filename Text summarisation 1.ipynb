```python
def cleanup(doc) -> list:
    tokens = [tok for tok in doc if not tok.is_punct and not tok.is_stop and not tok.is_space]
    return tokens
```

```python
def word_frequency(tokens) -> dict:
    word_frequency = {}
    
    for tok in tokens:
        if tok.text in word_frequency.keys():
            word_frequency[tok.text] += 1
        else:
            word_frequency[tok.text] = 1

    max_frequency = max(word_frequency.values())

    for key in word_frequency.keys():
        word_frequency[key] = word_frequency[key] / max_frequency
    
    return word_frequency
```

```python
def sentence_score(doc, word_frequency: dict) -> dict:
    sentence_score = {}
    for sent in doc.sents:
        sentence_score[sent] = 0
        for token in sent:
            if token.text in word_frequency.keys():
                sentence_score[sent] += word_frequency[token.text]
    return sentence_score
```

```python
from heapq import nlargest

def gen_summary(sentence_score: dict, summary_percentage = 30) -> str:
    sentence_position = dict(zip(sentence_score.keys(), range(0, len(sentence_score) -1)))
    summary_len = round(len(sentence_score) * summary_percentage / 100)
    summary_sents = nlargest(summary_len, sentence_score, key=sentence_score.get)
    summary_sents = sorted(summary_sents, key = lambda x: sentence_position[x])
    summary = " ".join(list(map(lambda s: s.text, summary_sents)))
    return summary
```

```python
def print_summary(text: str):
    doc = nlp(text)
    tokens = cleanup(doc)
    word_freq = word_frequency(tokens)
    sent_score = sentence_score(doc, word_freq)
    summary = gen_summary(sent_score)
    print(summary)
    print("\nSummary details: \nOriginal content length : %s \nSummary length : %s" % (len(text), len(summary)))
```

```python
print_summary(text)
```

```python
# nlargest from an array
arr = [1, 2, 6, 3, 5, 19, 15, 16, 12, 4]
nlargest(3, arr)
```

```python
ziplist = dict(zip(arr, range(0, len(arr)-1)))
largest = nlargest(3, arr)
top_list = list(map(lambda x: [x, ziplist[x]], largest))
print(top_list)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# simple linear function f(x,y) = 3x + 2y - 2 
g = lambda x, y: 3*x + 2*y - 2
```

```python
xs = np.linspace(-10, 10, 100)
fx = lambda x: (2 - 3*x) / 2.0
ys = fx(xs)
```

```python
# the function g's solution represent a line which divides the xy plane into two parts
# how do we predict a point p lies below or above the this line ?
plt.plot(xs, ys)
```

```python
# for any point above the line the function returns a +ve value

f(5, 5)
```

```python
# for any point below the line the function retuns -ve value
f(-2.5, -5)
```

```python
# Lets create a list of random points which are going to be our inputs X

ix, iy = np.linspace(-10, 10, 100), np.linspace(-10, 10, 100)
np.random.shuffle(ix)
np.random.shuffle(iy)
X = np.array(list(zip(ix, iy)))

# now lets see which of all these points are above the line and on or below the line
# we say 1 means above the line 0 means on or below the line that is going to be our 
# results Y
Y = f(X[:, 0], X[:, 1])
```

```python
fig, ax = plt.subplots()
ax.plot(xs, ys)
ax.scatter(X[:, 0], X[:, 1])
plt.show()
```

```python
above_or_below = lambda x: (x > 0).astype(int)
```

```python
Y = above_or_below(Y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=2)
```

### Learning rate decay

```python
alpha = 1.0
decay_rate = 0.1
```

```python
lr_f = lambda i: alpha / (1 + decay_rate * i)
```

```python
xs = np.arange(0, 10000, 100) + 1
ys = lr_f(xs)
```

```python
plt.plot(xs, ys)
```

### cache decay 

```python
rho = 0.9
```

```python
def cache_f(grad, cache):
    cache = (rho*cache) + (1-rho)*(grad**2) 
    return cache
```

```python
gradF = lambda x: 2*x
```

```python
xs = np.linspace(0,2,10000)
cache_hist = []
cache = 0

for x in xs:
    grad = gradF(x)
    cache = cache_f(grad, cache)
    cache_hist.append(cache)
```

```python
alpha = 1.0
def ada_grad_lre(cache, epsilon=1e-5):
    return alpha / (np.sqrt(cache) + epsilon)
```

```python
plt.plot(range(len(cache_hist)), cache_hist)
```

```python
lre = ada_grad_lre(cache_hist)
```

```python
plt.plot(range(len(lre[100:])), lre[100:])
```

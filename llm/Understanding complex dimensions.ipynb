```python
import torch 
```

```python
b_size = 1
n_tokens = 2
dim_in = 6
dim_out = 6
n_heads = 3
dim_head = 2
```

```python
torch.manual_seed(123)
```

```python
x = torch.rand(n_tokens, dim_in).unsqueeze(dim=0)
```

```python
x.shape # batch containing 1 item with 2x6 
```

```python
x
```

```python
y = x.view(b_size, n_tokens, n_heads, dim_head)
y
```

```python
y.shape
```

```python
keys = y.transpose(1,2)
```

```python
keys
```

```python
keys_t = keys.transpose(2,3)
```

```python
keys_t
```

```python
attn_scores = keys @ keys_t
```

```python
attn_scores
```

```python
mask = torch.triu(torch.ones(n_tokens, n_tokens), diagonal=1)
mask
```

```python
attn_scores.masked_fill(mask.bool()[:n_tokens, :n_tokens], -torch.inf)
```

```python
start = df["PRICE"].min()
```

```python
window = df["PRICE"].std().astype(int)
```

```python
value_counts_sorted = df["PRICE"].value_counts(bins=range(start, end, window))
```

```python
higest_range = value_counts_sorted[value_counts_sorted == value_counts_sorted.max()].index
```

```python
most_resistent_price = higest_range[0].mid
```

```python
from pathlib import Path
```

```python
output_file = Path("/home/deepak/Data/smc-corpus/graphemes.txt")
total_graphemes = 0
graphemes = set()

for file in list_files("/home/deepak/Data/smc-corpus/text", pattern="*.txt"):
    with file.open(encoding="utf-8") as f:
        text = f.read()
    normalized_text = my_normalizer.normalize_str(text)
    graphemes_in_text = set(split_by_grapheme(normalized_text))
    graphemes = graphemes.union(graphemes_in_text)
    no_graphemes = len(graphemes_in_text)
    total_graphemes += no_graphemes
    print(no_graphemes)

print(f"Total {total_graphemes}\n", sorted(list(graphemes)))

with output_file.open("w", encoding="utf-8") as out:
    out.write("\n".join(sorted(list(graphemes))))
```

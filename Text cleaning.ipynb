```python
import re
import string
import sys
import os

def remove_non_ascii(a_str):
    ascii_chars = set(string.printable)

    return ''.join(
        filter(lambda x: x in ascii_chars, a_str)
    )

def clean_text(text):
    step1 = remove_non_ascii(text)
    step2 = re.sub(r'[\r\n][\r\n]{2,}', '\n\n', step1)
    step3 = re.sub(r'\s+,', ",", step2) # remove space before ,
    step4 = re.sub(r'\s+\.,', ".", step3) # remove space before .
    step5 = step4.replace(":", " : ") # add space around :
    step6 = re.sub(r' +', ' ', step5) # remove duplicate spaces
    step7 = "\n".join([line.strip() for line in step6.split("\n")])
    return step7
```

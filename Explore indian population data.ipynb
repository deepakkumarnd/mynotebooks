```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.style.use("ggplot")
```

```python
df = pd.read_csv("./data/2011-IndiaState.csv")
```

```python
df.columns
```

```python
df.head(10)
```

```python
df = df[['Level','Name', 'TRU', 'No_HH', 'TOT_P', 'TOT_M', 'TOT_F',
        'P_LIT','M_LIT', 'F_LIT', 'P_ILL', 'M_ILL', 'F_ILL', 
        'TOT_WORK_P', 'TOT_WORK_M','TOT_WORK_F', 
        'MAINWORK_P', 'MAINWORK_M', 'MAINWORK_F', 'NON_WORK_P',
       'NON_WORK_M', 'NON_WORK_F']].copy()
```

```python
df.head()
```

### Top populated indian states

```python
df.query('Level == "STATE" and TRU == "Total"').nlargest(10, "TOT_P").plot("Name", "TOT_P", kind="barh")
plt.show()
```

#### Sex ratio

```python
df["sexratio"] = df["TOT_F"] / df["TOT_M"]
```

```python
df.query('Level == "STATE" and TRU == "Total"').nlargest(10, "sexratio")[["Name", "sexratio"]].plot("Name", "sexratio", kind="barh")
```

#### Working population

```python
df["working_percentage"] = df["TOT_WORK_P"]/df["TOT_P"]*100
df["non_working_percentage"] = df["NON_WORK_P"]/df["TOT_P"]*100
```

```python
df.query('Level == "STATE" and TRU == "Total"').nlargest(10, "working_percentage")[["Name", "working_percentage"]].plot("Name", "working_percentage", kind="barh")
```

```python
df.query('Level == "STATE" and TRU == "Total"').nlargest(10, "non_working_percentage")[["Name", "non_working_percentage"]]
```

```python
df["female_working_percentage"] = df["TOT_WORK_F"]/df["TOT_F"]*100
df.query('Level == "STATE" and TRU == "Total"').nlargest(10, "female_working_percentage")[["Name", "female_working_percentage"]]
```

```python
df["male_working_percentage"] = df["TOT_WORK_M"]/df["TOT_M"]*100
df.query('Level == "STATE" and TRU == "Total"').nlargest(10, "male_working_percentage")[["Name", "male_working_percentage"]]
```

#### literacy


```python
df["lit_percentage"] = df["P_LIT"]/df["TOT_P"] * 100
```

```python
df.query('Level == "STATE" and TRU == "Total"')\
.nlargest(10, "lit_percentage")[["Name", "lit_percentage"]]
```

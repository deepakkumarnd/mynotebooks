```python
from spacy.lang.en import English

def extract_mobile(text):
    nlp = English()
    ruler = nlp.add_pipe("entity_ruler")
    patterns = [{"label": "IND_PHONE1", "pattern": [{"TEXT": "+91"}, { "OP" : "?"}, {"SHAPE": "dddd", "LENGTH": 10}]},
                {"label": "IND_PHONE2", "pattern": [{"TEXT": "+91"}, { "OP" : "?"}, {"SHAPE": "dddd", "LENGTH": 5}, { "OP" : "?"}, {"SHAPE": "dddd", "LENGTH": 5}]},
                {"label": "IND_PHONE3", "pattern": [{"SHAPE": "dddd", "LENGTH": 11}]},
                {"label": "IND_PHONE4", "pattern": [{"TEXT": "0"}, { "OP" : "?"}, {"SHAPE": "dddd", "LENGTH": 10}]}
               ]
    ruler.add_patterns(patterns)

    doc = nlp(text)
    print([(ent.text, ent.label_) for ent in doc.ents])
```

```python
extract_mobile(text1)
```

```python
from spacy.matcher import Matcher

def extract_email(text):
    matcher = Matcher(nlp.vocab)
    matcher.add("emailMatcher", [[{"LIKE_EMAIL": True}]])
    doc = nlp(text)
    matches = matcher(doc)
    for match_id, start, end in matches:
        span = doc[start:end]
        print(span.text)
```

```python
extract_email(text)
```

```python
def extract_link(text):
    matcher = Matcher(nlp.vocab)
    matcher.add("linkMatcher", [[{"LIKE_URL": True}]])
    doc = nlp(text)
    matches = matcher(doc)
    for match_id, start, end in matches:
        span = doc[start:end]
        print(span.text)
```

```python
def extract_names(text):
    doc = nlp(text)
    for ent in doc.ents:
        if ent.label_ == "PERSON":
            print(ent.text)
```

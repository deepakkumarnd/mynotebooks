```python
# generate n levels of pascals triangle

def pascals_triangle(level, padding=100):
    if level <= 1:
        print(" ".join(map(str, [1])).center(padding), "Sum = ", 1)
        return [1]

    previous_level = pascals_triangle(level - 1)
    l = len(previous_level)
    mid = int(l / 2)
    previous_level.insert(mid + 1, previous_level[mid]) # double the middle coefficient
    previous_level.insert(0, 0)
    previous_level.append(0)

    new_level = [0] * (l + 2)
    for i, pair in enumerate(zip(previous_level, previous_level[1:])):
        a, b = pair
        new_level[i] = a + b

    print(" ".join(map(str, new_level)).center(padding), "Sum = ", sum(new_level), "Mean = {0:.2f}".format(sum(new_level) / len(new_level)))
    return new_level
```

```python
pascals_triangle(10)
```

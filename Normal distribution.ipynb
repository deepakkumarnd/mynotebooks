```python
import numpy as np
```

```python
random_numbers = np.random.rand(10) * 100
print(random_numbers)
```

```python
np.var(random_numbers)
```

```python
mean = np.mean(random_numbers)
```

```python
mean
```

```python
normalised = (random_numbers - mean) / np.sqrt(np.var(random_numbers))
```

```python
normalised.mean(),  normalised.var()
```
